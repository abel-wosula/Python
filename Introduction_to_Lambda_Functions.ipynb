**Lambda Functions in Python**

**Intro**

Is a one line anonymous function.
Most useful when being used as an argument.

we use lambda instead of def when creating a lamda function.

the function body can only have one expression.



Strucure



```python
add_1 = lambda x: x+1
print(add_1(9))
```

```python
# can accept more than 1 paramenter and returns any valid Python expression.

add_2 = lambda x, y: x + y
print(add_2(3,6))
```

Use cases
1. using the Map function.
2. Using filter function
3. Using the Sorted function

```python
# map function to apply an operable on an list.
my_list = [1,2,3,4,5]

squares = list(map(lambda x: x ** 2, my_list))
print(squares)
```

```python
# filter function on a list

evens = list(filter(lambda x: x % 2 == 0, my_list))
print(evens)
```

```python
values = [(1, 'b', 'hello'), (2, 'a', 'world'), (3, 'c', 'python')]
sorted_values = sorted(values, key=lambda x: x[2])
print(list(sorted_values))
```

```python
# Advanced Summing technique!
from functools import reduce

my_list = [1,2,3,4,5]

sum_of_numbers = reduce(lambda x, y: x + y, my_list)
print(sum_of_numbers)

#Finding the max value
max_value = reduce(lambda x, y: x if x > y else y, my_list)
print(max_value)

# Finding the minimum value
min_value = reduce(lambda x, y: x if x < y else y, my_list)
print(min_value)
```
