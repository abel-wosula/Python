## ALGORITHMS AND ALGORITHMIC COMPLEXITY

An algorithm is a step-by-step procedure or formula for solving a problem. Algorithms are fundamental to computer science and are used in various applications, from sorting data to searching databases.

###  Understanding Algorithmic Complexity and Big O Notation
Algorithmic complexity refers to the amount of resources an algorithm consumes as a function of the size of the input. The most common resource considered is time (time complexity), but space (space complexity) is also important. Big O notation is used to classify algorithms according to how their run time or space requirements grow as the input size grows.

 Time Complexity refers to the computational complexity that describes the amount of time it takes to run an algorithm as a function of the length of the input. Common time complexities include:
- O(c): Constant     
- O(log n): Logarithmic time
- O(n): Linear time
- O(n log n): Linearithmic time
- O(n^2): Quadratic time

 Space Complexity refers to the amount of memory space an algorithm uses in relation to the input size. Similar to time complexity, space complexity can also be expressed using Big O notation.

```python
# Case example
# Python program to calculate factorial of a number - approach 1
def fact(n):
    product = 1
    for i in range(n):
        product = product * (i+1)
    return product

print(fact(5))
```

```python
# Python program to calculate factorial of a number - approach 2 (Recursion)
def fact2(n):
    if n == 0:
        return 1
    else:
        return n * fact2(n-1)

print(fact2(5))
```

Finding the time required to execute code of the same output.

```python
%timeit fact(50)
```

```python
%timeit fact2(50)
```

In large inputs the difference is significant.
The Big-O notation signifies the relationship between the input to the algorithm and the steps required to execute the algorithm.

The key takeaway is - the Big-O isn't interested in a particular instance in which you run an algorithm, such as fact(50), but rather, in how well it scales given increasing input. This is a much better metric for evaluating than concrete time for a concrete instance!

To build intuition:

O(n): at n=1, 1 step is taken.
At n=10, 10 steps are taken.

O(n²): at n=1, 1 step is taken.
At n=10, 100 steps are taken.

At n=1, these two would perform the same! This is another reason why observing the relationship between the input and the number of steps to process that input is better than just evaluating functions with some concrete input.

Generally, anything worse than linear is considered a bad complexity (i.e. inefficient) and should be avoided if possible. Linear complexity is okay and usually a necessary evil. Logarithmic is good. Constant is amazing!



Constant Complexity - Steps required to complete the execution of an algorithm remain constant, irrespective of the number of inputs. Denoted by O(c) where c is a constant number.

```python
# Constant complexity in Python
# program that finds the square of the first item in a list and then prints it on the screen.

def constant_algo(items):
  results = items[0] * items[0]
  print(results)

constant_algo([4,5,5,7,8])
```

Linear Complexity - Steps required to complete the execution of an algorithm increase or decrease linearly with the number of inputs. Denoted by O(n).

```python
# Linear complexity in Python
# Programs that prints all the items in a list.

def linear_algo(items):
  for item in items:
    print(item)
linear_algo([3,3,4,6])
```

Quadratic Complexity - steps required to execute an algorithm are a quadratic function of the number of items in the input. Denoted as O(n^2).

```python
# Quadratic Complexity in Python
# Program that prints nested items in loops.

def quadratic_algo(items):
  for item in items:
    for item2 in items:
      print(item, " ", item2)
quadratic_algo([4,9,0])
```

Logarithmic Complexity - An algorithm exhibits logarithmic complexity when it reduces the problem size by a constant factor in each step, such as dividing the input in half. Denoted by O(log n).
Common in binary Search.

```python
# Logarithmic Complexity in Python
# Program performs binary search.
# Divide and conquer algorithm.
def binary_search(arr, target):
   left, right = 0, len(arr) - 1
   while left <= right:
       mid = (left + right) // 2
       if arr[mid] == target:
           return mid
       elif arr[mid] < target:
           left = mid + 1
       else:
           right = mid - 1
   return -1
# Input size halves with each iteration
print(binary_search([1, 2, 3, 4, 5, 6, 7, 8], 6)) # O(log n)
```
