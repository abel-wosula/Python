Search Algorithms - Linear Search & Binary Search.

Core Problem - we need to determine if a particular character, string or number is present within a data structure.



```python
#We can choose to use In-built Python methods

# For strings - use find() method, or RegEx

my_string = "Allan Joseph"

print(my_string.find("e"))

# For lists use the in keyword

my_list = [1,5,6,"four","apple", "pear","grape"]

if "apple" in my_list:
  print("Apple is in the list")

```

However, if the data structure is large or the search use case is more complex, we would need searching algorithms with greater flexibility and efficiency.



Linear Search

The algorithm begins at one end of the list (or data structure) and checks each element until it finds the correct element or it reaches the end of the list.

The algorithm terminates when it finds an occurrence of the searched element.



```python
# Pseudocode
# procedure linear_search( list, target )
#     for each element in the list
#         if the element equals the target
#             return its index

#     return not found
# end procedure
```

**Time Complexity**

The highest overall level of either average or worst-case time complexity is linear time,  O(n)
 , because as the input size  n
  increases, the algorithm run-time scales with the input size. The best-case time complexity is then constant time,  O(1)
 , when the searched item is the first element in the list.

 As a result, linear search is not used often because the same efficiency may be achieved with Python's built-in search techniques. However, a linear search algorithm is useful to find the first occurrence of an element in an unsorted collection, unlike most other search algorithms that can only be applied to sorted collections.
